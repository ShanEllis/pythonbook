Provided function:
```python
def convert_temp(temp, input_unit='F'):
    
    if input_unit == 'F':
        out_temp = (temp - 32) * 5/9
    elif input_unit == 'C':
        out_temp = (temp * 9/5) + 32
    else:
        print("Please specify either 'F' or 'C' for input_unit")
        
    return out_temp
```

Checks you can use to see if you're on the right track:

````python
assert tempA == 35.0
assert tempB == 62.6
````

