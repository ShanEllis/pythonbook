## Documentation

### Documentation Q1

```python
# exact docstring and comment wording will differ
def count_evens(number_list):
    """Return the number of even values in the input list.

    Parameters
    ----------
    number_list : list of int
        The list of numbers to check for even values.

    Returns
    -------
    count : int
        The number of even values in number_list.
    """

    count = 0

    for value in number_list:
        # a value is even if dividing it by 2 leaves no remainder
        if value % 2 == 0:
            count += 1

    return count
```

## Code Testing

### Code Testing Q1

```python
class Person:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def birthday(self):
        self.age += 1
        return 'Happy Birthday!'
```

```python
import unittest

class TestPerson(unittest.TestCase):

    def setUp(self):
        # runs before EACH test method - every test gets a fresh Person
        self.jonathan = Person('Jonathan', 20)

    def test_is_person_instance(self):
        self.assertIsInstance(self.jonathan, Person)

    def test_attributes(self):
        self.assertEqual(self.jonathan.name, 'Jonathan')
        self.assertEqual(self.jonathan.age, 20)

    def test_birthday_increments_age(self):
        output = self.jonathan.birthday()
        self.assertEqual(self.jonathan.age, 21)
        self.assertEqual(output, 'Happy Birthday!')

unittest.main(argv=['', 'TestPerson'], verbosity=2, exit=False);
```
